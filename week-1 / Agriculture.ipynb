```python
crop_data={
    "crop_name":"Rice",
    "Scientic_name":"Oryza sativa",
    "crop_type":"Cereal",
    "season":"Kharif",
    "ideal_temperature_celsius":25,
    "annual_rainfall_mm": 1200,
    "soil_type":"Clayey Loam",
    "ideal_soil-ph": 6.5,
    "growing_duration_days": 120,
    "water_requirement":"High",
    "average_yield_per_hectare_kg":4000,
    "primary_nutrient":["Nitrogen","Phosphorus","Potassium"],
    "common_pests":["Stem Borer","Brown Planthopper"],
    "harvesting_method":"Manual/Mechanical",
    "storage_humidity_percent":13
}
print(crop_data)
crop_data["Area"]="ELuru"
crop_data["Pesticide"]="Urea"

print(crop_data)

crop_data["primary_nutrients"]=["Sulfur (S)","Zinc (Zn)","Magnesium (Mg)","Iron (Fe)"]
crop_data["common_pests"]=["Aphids","Spider mites","Brown whitefiles","Thips","Cutworms"]
print(crop_data)
```

```python
import pandas as pd

# Format list values as strings so they store cleanly in a single CSV cell
formatted_data = {}
for key, value in crop_data.items():
    if isinstance(value, list):
        formatted_data[key] = [", ".join(value)]
    else:
        formatted_data[key] = [value]

# Convert to DataFrame and save to CSV
df = pd.DataFrame(formatted_data)
df.to_csv('crop_data.csv', index=False)

print("Saved crop_data.csv successfully!")
display(df)
```
